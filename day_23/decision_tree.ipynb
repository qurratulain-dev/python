```python
import seaborn as sns

df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
df.info()
```

# label encoder wala is liye kr ry h q k smoker ki type category h 

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
le.fit(df['smoker'])

df['smoker'] = le.transform(df['smoker'])

# for day
le = LabelEncoder()
le.fit(df['day'])

df['day'] = le.transform(df['day'])

# for dinner
le = LabelEncoder()
le.fit(df['time'])

df['time'] = le.transform(df['time'])
```

```python
df.isnull().sum()
```

```python
X = df.drop(columns=['sex'])
y= df['sex']
```

```python
print(X.head())
print(y.head())
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model = model.fit(X,y)
```

```python
import joblib
joblib.dump(model, 'decision_tree_tipswalamodel.joblib')
```

```python
model=joblib.load('decision_tree_tipswalamodel.joblib')
```
