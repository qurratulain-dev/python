```python
import pandas as pd
import requests
from io import StringIO
```

```python
# ---------------------------------------------------------
# 1. Local CSV File Load Karna
# ---------------------------------------------------------
df = pd.read_csv('aug_train.csv')
print(df.head())
```

```python
# ---------------------------------------------------------
# 2. URL se Direct CSV File Load Karna
# ---------------------------------------------------------
url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)
df_url = pd.read_csv(data)
```

```python
# ---------------------------------------------------------
# 3. Sep Parameter (Separator)
# Jab file comma (,) se na, balki tab (\t) ya semicolon (;) se separated ho
# ---------------------------------------------------------
df_sep = pd.read_csv('movie_titles_metadata.tsv', sep='\t', 
                     names=['sno', 'name', 'release_year', 'rating', 'votes', 'genres'])
```

```python
# ---------------------------------------------------------
# 4. Index_col Parameter
# Kisi specific column ko DataFrame ka index bana dene ke liye
# ---------------------------------------------------------
df_index = pd.read_csv('aug_train.csv', index_col='enrollee_id')
```

```python
# ---------------------------------------------------------
# 5. Header Parameter
# Agar column names pehli row mein na hon, ya kisi aur row mein hon
# ---------------------------------------------------------
df_header = pd.read_csv('test.csv', header=1) # Row index 1 ko header bana dega
```

```python
# ---------------------------------------------------------
# 6. Usecols Parameter
# Sirf specific columns import karne ke liye (Memory bachane ke liye best)
# ---------------------------------------------------------
df_usecols = pd.read_csv('aug_train.csv', usecols=['enrollee_id', 'gender', 'education_level'])
```

```python
# ---------------------------------------------------------
# 7. Squeeze Parameter
# Agar sirf 1 column import karna ho aur usay DataFrame ki jagah Series banana ho
# ---------------------------------------------------------
s_series = pd.read_csv('aug_train.csv', usecols=['gender'])['gender']
```

```python
# ---------------------------------------------------------
# 8. Nrows / Skiprows Parameter
# Limited rows import karna ya beech ki rows skip karna
# ---------------------------------------------------------
df_nrows = pd.read_csv('aug_train.csv', nrows=100)       # Sirf pehli 100 rows layega
df_skip = pd.read_csv('aug_train.csv', skiprows=[0, 2])   # Row 0 aur 2 ko skip kar dega
```

```python
# ---------------------------------------------------------
# 9. Encoding Parameter
# Agar file UTF-8 mein na ho (jaise Zomato dataset aksar 'latin-1' mein hota hai)
# ---------------------------------------------------------
df_encode = pd.read_csv('zomato.csv', encoding='latin-1')
```

```python
# ---------------------------------------------------------
# 10. Bad Lines Handle Karna
# Agar kisi row mein columns zyada hon (Parser Error), to usay skip kar do
# Note: Naye Pandas versions mein 'error_bad_lines' ki jagah 'on_bad_lines' use hota hai.
# ---------------------------------------------------------
df_bad = pd.read_csv('BX-Books.csv', sep=';', encoding="latin-1", on_bad_lines='skip')
```

```python
# ---------------------------------------------------------
# 11. Dtype Parameter
# Memory bachane ke liye column ka data type import time par hi change karna
# ---------------------------------------------------------
df_dtype = pd.read_csv('aug_train.csv', dtype={'target': int})
```

```python
# ---------------------------------------------------------
# 12. Parse Dates Parameter
# Date column ko string ki jagah proper 'datetime' object mein convert karna
# ---------------------------------------------------------
df_dates = pd.read_csv('IPL Matches 2008-2020.csv', parse_dates=['date'])
```

# convertrs

```python
# ---------------------------------------------------------
# 13. Converters Parameter
# Data load karte waqt hi kisi custom function se value change karna
# ---------------------------------------------------------
def rename_team(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    return name

df_convert = pd.read_csv('IPL Matches 2008-2020.csv', converters={'team1': rename_team})
```

```python
# ---------------------------------------------------------
# 14. NA_Values Parameter
# Custom values (jaise '-', 'NA', ya 'Male') ko NaN (Missing Value) treat karna
# ---------------------------------------------------------
df_na = pd.read_csv('aug_train.csv', na_values=['-', 'Unknown'])
```
