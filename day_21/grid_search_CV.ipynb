```python
# Linear Regression
import pandas as pd
import seaborn as sns
df = sns.load_dataset("titanic")

X = df[['age']]
y = df['fare']

# NaN values ko mean se fill karna
X = X.fillna(X.mean()) 

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV

# create a model
model = LinearRegression()

# define parameter grid (normalize hata diya hai kyunke naye version mein support nahi karta)
grid_parameter = {'fit_intercept': [True, False]}

# object grid search cv
grid_search = GridSearchCV(model, grid_parameter, cv=5, scoring='r2')

# train the model
grid_search.fit(X, y)

# print best parameters
print("best parameters: ", grid_search.best_params_)

# Prediction (grid_search use karein, model nahi)
y_pred = grid_search.predict([[25]])
print("Predicted Fare for age 25:", y_pred)
```

# KNN Grid Search cv

```python
# K-Nearest Neighbors Classification (Pehle Linear Regression likha tha, wo change kar diya)
import pandas as pd
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings('ignore') # Warning messages ko hide karne ke liye

df = sns.load_dataset("titanic")

X = df[['pclass','sex', 'age', 'sibsp', 'parch','fare']]
y = df['survived']

# 'sex' column ko 0 aur 1 (dummy variables) mein convert karna
X = pd.get_dummies(X, columns=['sex'], drop_first=True)

# NaN values ko mean se fill karna
X = X.fillna(X.mean()) 

from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

# create a model
model = KNeighborsClassifier()

# define parameter grid
grid_parameter = {'n_neighbors': np.arange(1,40,2), 'weights': ['uniform', 'distance']}
# object grid search cv
grid_search = GridSearchCV(model, grid_parameter, cv=5, scoring='recall')

# train the model
grid_search.fit(X, y)

# print best parameters and score
print("best parameters: ", grid_search.best_params_)
print("best Score: ", grid_search.best_score_)

# --- Prediction Section (Corrected) ---

# get_dummies use karne ke baad columns ka order yeh banta hai: 
# ['pclass', 'age', 'sibsp', 'parch', 'fare', 'sex_male']
# Isliye humein 6 values deni hongi.
# Example: Pclass=3, Age=25, SibSp=1, Parch=0, Fare=10.0, Sex=Male(1)
passenger_data = [[3, 25, 1, 0, 10.0, 1]] 

# Prediction (grid_search use karein)
y_pred = grid_search.predict(passenger_data)

# Print statement ko bhi theek kar diya
print("Predicted Survival (0=No, 1=Yes):", y_pred)
```

# DT C

```python
# Decision Tree Classifier (DTC)
import pandas as pd
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings('ignore') # Warning messages ko hide karne ke liye

df = sns.load_dataset("titanic")

X = df[['pclass','sex', 'age', 'sibsp', 'parch','fare']]
y = df['survived']

# 'sex' column ko 0 aur 1 (dummy variables) mein convert karna
X = pd.get_dummies(X, columns=['sex'], drop_first=True)

# NaN values ko mean se fill karna
X = X.fillna(X.mean()) 

# DTC import karna
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

# create a model (random_state add kiya hai taake results consistent rahein)
model = DecisionTreeClassifier(random_state=42)

# define parameter grid for Decision Tree
grid_parameter = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [None, 5, 10, 15, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# object grid search cv (scoring='accuracy' rakha hai, aap 'f1' ya 'recall' bhi use kar sakte hain)
grid_search = GridSearchCV(model, grid_parameter, cv=5, scoring='accuracy')

# train the model
grid_search.fit(X, y)

# print best parameters and score
print("best parameters: ", grid_search.best_params_)
print("best Score: ", grid_search.best_score_)

# --- Prediction Section ---

# get_dummies use karne ke baad columns ka order yeh banta hai: 
# ['pclass', 'age', 'sibsp', 'parch', 'fare', 'sex_male']
# Example: Pclass=3, Age=25, SibSp=1, Parch=0, Fare=10.0, Sex=Male(1)
passenger_data = [[3, 25, 1, 0, 10.0, 1]] 

# Prediction (grid_search use karein)
y_pred = grid_search.predict(passenger_data)

# Print statement
print("Predicted Survival (0=No, 1=Yes):", y_pred)
```
